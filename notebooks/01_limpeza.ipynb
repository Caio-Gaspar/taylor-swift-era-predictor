```python
import pandas as pd

df_principal = pd.read_csv('../dados/taylor_swift_spotify.csv')
df_showgirl = pd.read_csv('../dados/showgirl.csv')

df_completo = pd.concat([df_principal, df_showgirl], ignore_index=True)
df_completo.to_csv('../dados/taylor_swift_completo.csv', index=False)
df = df_completo

print(f"Concatenação concluída! Total de faixas: {len(df)}")
df.tail(12)
```

```python
if 'speachiness' in df.columns:
    df.rename(columns={'speachiness': 'speechiness'}, inplace=True)
```

Remoção de faixas avulsas, singles e feats, e repetidas por conta de álbuns deluxe ou taylor's version

```python
albuns_oficiais = [
    'Taylor Swift', 
    "Fearless (Taylor's Version)", 
    "Speak Now (Taylor's Version)",
    "Red (Taylor's Version)",
    "1989 (Taylor's Version)",
    "reputation",
    "Lover",
    "folklore (deluxe version)",
    "evermore (deluxe version)",
    "Midnights (The Til Dawn Edition)",
    "THE TORTURED POETS DEPARTMENT: THE ANTHOLOGY",
    "The Life of a Showgirl"
]
df_limpo = df[df['album_name'].isin(albuns_oficiais)].copy()
print(f"Filtragem concluída! Total de faixas após filtragem: {len(df_limpo)}")
```

```python
termos_remover = ["remix", "live", "demo", "acoustic", "instrumental", "karaoke"]
padrao_regex = '|'.join(termos_remover)
df_limpo = df_limpo[~df_limpo['track_name'].str.contains(padrao_regex, case=False, na=False)]
print(f"Remoção de faixas com termos indesejados concluída! Total de faixas após remoção: {len(df_limpo)}")
```
